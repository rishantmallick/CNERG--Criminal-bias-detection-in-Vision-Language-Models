```python
from PIL import Image
import torch
from transformers import CLIPProcessor, CLIPModel
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch16")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch16")
```

```python
text_prompts = [
    "a man with a beard",
    "a man with no beard",
    "a man with a moustache",
    "a man without a moustache",
    "a person with a round face",
    "a person with a slim face",
    "a messy looking person",
    "a well groomed person"
]

face_dictionary = {
    "a man with a beard":[],
    "a man with no beard":[],
    "a man with a moustache":[],
    "a man without a moustache":[],
    "a person with a round face":[],
    "a person with a slim face":[],
    "a messy looking person":[],
    "a well groomed person":[]
}
```

```python
for i in range(0,5000):
    test_img_path = f"/content/drive/MyDrive/criminals/rajasthanmetadata/{i:05}.jpg"
    image = Image.open(test_img_path)
    image = image.convert("RGB")
    print(f"{i:05}.jpg is done")
    inputs = processor(
       text=text_prompts, images=image, return_tensors="pt", padding=True
     )
    with torch.inference_mode():
      outputs = model(**inputs)
    logits_per_image = outputs.logits_per_image  # this is the image-text similarity score
    probs = logits_per_image.softmax(dim=1)
    print(probs)
    if probs[0][0] > probs[0][1]:
      face_dictionary["a man with a beard"].append(f"{i:05}.jpg")
    else:
      face_dictionary["a man with no beard"].append(f"{i:05}.jpg")
    if probs[0][2] > probs[0][3]:
      face_dictionary["a man with a moustache"].append(f"{i:05}.jpg")
    else:
      face_dictionary["a man without a moustache"].append(f"{i:05}.jpg")
    if probs[0][4] > probs[0][5]:
      face_dictionary["a person with a round face"].append(f"{i:05}.jpg")
    else:
      face_dictionary["a person with a slim face"].append(f"{i:05}.jpg")
    if probs[0][6] > probs[0][7]:
      face_dictionary["a messy looking person"].append(f"{i:05}.jpg")
    else:
      face_dictionary["a well groomed person"].append(f"{i:05}.jpg")
```
